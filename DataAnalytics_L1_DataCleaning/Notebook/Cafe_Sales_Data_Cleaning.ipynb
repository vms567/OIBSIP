## Data Cleaning

### Handling Invalid Values

The dataset contains invalid placeholder values such as `ERROR` and `UNKNOWN`. These values are converted to missing values (`NaN`) so that appropriate cleaning strategies can be applied to each column.

```python
# Replace invalid values with NaN
df.replace(['ERROR', 'UNKNOWN'], np.nan, inplace=True)

# Check missing values after replacement
df.isnull().sum()
```

```python
df
```

### Correcting Numeric Data Types

The Quantity, Price Per Unit, and Total Spent columns were stored as object data types because they originally contained invalid text values. After replacing these invalid values with missing values, the columns are converted to numeric format.

```python
# Convert columns to numeric
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'], errors='coerce')
df['Total Spent'] = pd.to_numeric(df['Total Spent'], errors='coerce')

# Check data types
df.dtypes
```

The Transaction Date column was stored as an object data type. Since it represents dates, it is converted to datetime format for accurate date-based analysis.

```python
# Convert Transaction Date to datetime
df['Transaction Date'] = pd.to_datetime(
    df['Transaction Date'],
    format='%d-%m-%Y',
    errors='coerce'
)

df['Transaction Date'].head()
```

## Handling Missing Values

### Missing Item Values

Before filling missing values in the Item column, the relationship between Item and Price Per Unit is examined. If individual items have consistent prices, the price information may help identify missing item names.

```python
# Check unique prices for each item
df.groupby('Item')['Price Per Unit'].unique()
```

```python
df.isnull().sum()
```

### Handling Missing Item Values

Missing Item values cannot be reliably inferred because some products share the same unit price. Instead of deleting these transactions or guessing the product, missing Item values are labeled as `Unknown` to preserve the available transaction information.

```python
# Fill missing Item values
df['Item'] = df['Item'].fillna('Unknown')

# Check remaining missing Item values
df['Item'].isnull().sum()
```

```python
# Check missing values in numerical columns
df[['Quantity', 'Price Per Unit', 'Total Spent']].isnull().sum()
```

```python
# Check rows with missing numerical values
df[df[['Quantity', 'Price Per Unit', 'Total Spent']].isnull().any(axis=1)].head(10)
```

### Recovering Missing Numerical Values

The numerical columns are logically related:

**Total Spent = Quantity × Price Per Unit**

Therefore, missing numerical values are recovered using the available values wherever possible instead of immediately applying statistical imputation.

```python
# Fill missing Total Spent
df['Total Spent'] = df['Total Spent'].fillna(
    df['Quantity'] * df['Price Per Unit']
)

# Check remaining missing values
df['Total Spent'].isnull().sum()
```

```python
# Fill missing Quantity
df['Quantity'] = df['Quantity'].fillna(
    df['Total Spent'] / df['Price Per Unit']
)

# Check remaining missing values
df['Quantity'].isnull().sum()
```

```python
# Fill missing Price Per Unit
df['Price Per Unit'] = df['Price Per Unit'].fillna(
    df['Total Spent'] / df['Quantity']
)

# Check remaining missing values
df['Price Per Unit'].isnull().sum()
```

```python
# View remaining rows with missing numerical values
df[df[['Quantity', 'Price Per Unit', 'Total Spent']].isnull().any(axis=1)].head(20)
```

### Recovering Price Per Unit from Item

Each known cafe item has a fixed unit price. Therefore, remaining missing Price Per Unit values are recovered using the corresponding item's standard price wherever the item name is available.

```python
# Standard price for each item
item_prices = {
    'Cookie': 1.0,
    'Tea': 1.5,
    'Coffee': 2.0,
    'Cake': 3.0,
    'Juice': 3.0,
    'Sandwich': 4.0,
    'Smoothie': 4.0,
    'Salad': 5.0
}

# Fill missing Price Per Unit using Item
df['Price Per Unit'] = df['Price Per Unit'].fillna(
    df['Item'].map(item_prices)
)

# Check remaining missing values
df['Price Per Unit'].isnull().sum()
```

```python
# Fill remaining Quantity where possible
df['Quantity'] = df['Quantity'].fillna(
    df['Total Spent'] / df['Price Per Unit']
)

# Check remaining missing Quantity
df['Quantity'].isnull().sum()
```

```python
# Fill remaining Total Spent where possible
df['Total Spent'] = df['Total Spent'].fillna(
    df['Quantity'] * df['Price Per Unit']
)

# Check remaining missing Total Spent
df['Total Spent'].isnull().sum()
```

```python
# View remaining incomplete numerical records
df[df[['Quantity', 'Price Per Unit', 'Total Spent']].isnull().any(axis=1)]
```

### Removing Remaining Incomplete Numerical Records

Most missing numerical values were recovered using the relationship between Quantity, Price Per Unit, Total Spent, and Item.

However, a small number of records still contained insufficient information to reliably reconstruct the transaction. Since these records represent only a very small proportion of the dataset, they were removed rather than imputing potentially inaccurate numerical values.

```python
# Remove rows with remaining missing numerical values
df = df.dropna(
    subset=['Quantity', 'Price Per Unit', 'Total Spent']
).reset_index(drop=True)

# Check remaining missing numerical values
df[['Quantity', 'Price Per Unit', 'Total Spent']].isnull().sum()
```

### Handling Missing Payment Method

Payment Method contains a large number of missing values. Since payment method is a categorical attribute and the remaining transaction information is still useful, deleting these records would result in unnecessary data loss.

Therefore, missing Payment Method values are labeled as `Unknown`.

```python
# Fill missing Payment Method
df['Payment Method'] = df['Payment Method'].fillna('Unknown')

# Check missing values
df['Payment Method'].isnull().sum()
```

### Handling Missing Location

Location contains a large number of missing values. Since the transaction records still contain useful sales information, removing these rows would cause significant data loss.

Therefore, missing Location values are labeled as `Unknown`.

```python
# Fill missing Location
df['Location'] = df['Location'].fillna('Unknown')

# Check missing values
df['Location'].isnull().sum()
```

### Handling Missing Transaction Dates

Some Transaction Date values remained missing after invalid date entries were converted to missing values.

Since the actual transaction date cannot be reliably inferred from the other attributes, these records are removed rather than assigning artificial dates that could distort time-based analysis.

```python
# Remove rows with missing Transaction Date
df = df.dropna(subset=['Transaction Date']).reset_index(drop=True)

# Check missing Transaction Date values
df['Transaction Date'].isnull().sum()
```

## Outlier Detection and Treatment

Outliers are examined using the Interquartile Range (IQR) method. This method identifies values that fall significantly below or above the typical range of the data.

The Quantity column is examined first to determine whether unusually high or low transaction quantities are present.

```python
# Calculate IQR for Quantity
Q1 = df['Quantity'].quantile(0.25)
Q3 = df['Quantity'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# Count outliers
quantity_outliers = df[
    (df['Quantity'] < lower_limit) |
    (df['Quantity'] > upper_limit)
]

print("Number of Quantity Outliers:", len(quantity_outliers))
```

### Quantity Outlier Treatment

Using the IQR method, the lower and upper limits for Quantity were calculated as -1 and 7 respectively.

No Quantity values were identified outside these limits. Therefore, no outlier treatment was required for the Quantity column.

```python
# Calculate IQR for Price Per Unit
Q1 = df['Price Per Unit'].quantile(0.25)
Q3 = df['Price Per Unit'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# Count outliers
price_outliers = df[
    (df['Price Per Unit'] < lower_limit) |
    (df['Price Per Unit'] > upper_limit)
]

print("Number of Price Per Unit Outliers:", len(price_outliers))
```

### Price Per Unit Outlier Treatment

Using the IQR method, the lower and upper limits for Price Per Unit were calculated as -1 and 7 respectively.

No Price Per Unit values were identified outside these limits. Therefore, no outlier treatment was required for this column.

```python
# Calculate IQR for Total Spent
Q1 = df['Total Spent'].quantile(0.25)
Q3 = df['Total Spent'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# Count outliers
total_outliers = df[
    (df['Total Spent'] < lower_limit) |
    (df['Total Spent'] > upper_limit)
]

print("Number of Total Spent Outliers:", len(total_outliers))
```

```python
# View Total Spent outliers
total_outliers[
    ['Item', 'Quantity', 'Price Per Unit', 'Total Spent']
].head(20)
```

### Total Spent Outlier Treatment

The IQR method identified 259 Total Spent values outside the calculated upper limit of 24.

Further inspection showed that these values represent valid high-value transactions. For example, a quantity of 5 units at a price of 5 per unit correctly results in a Total Spent value of 25.

Since these values are logically consistent with the relationship:

**Total Spent = Quantity × Price Per Unit**

the detected outliers were retained rather than removed or capped. Removing them would result in the loss of valid transaction information.

```python
# Check Total Spent calculation
df['Calculated Total'] = df['Quantity'] * df['Price Per Unit']

# Find mismatched records
mismatch = df[df['Total Spent'] != df['Calculated Total']]

print("Number of mismatched records:", len(mismatch))
```

### Numerical Consistency Validation

The Total Spent column was validated against the relationship:

**Total Spent = Quantity × Price Per Unit**

After cleaning, no mismatched records were found. This confirms that the numerical transaction values are internally consistent.
